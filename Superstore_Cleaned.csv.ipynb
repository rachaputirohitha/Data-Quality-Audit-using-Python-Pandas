```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/Sample - Superstore.csv')

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.head()
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
missing_summary = pd.DataFrame({
    'Column': df.columns,
    'Missing Values': df.isnull().sum().values
})

missing_summary
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
print(df.dtypes)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')

print("Invalid Order Dates:", df['Order Date'].isnull().sum())
print("Invalid Ship Dates:", df['Ship Date'].isnull().sum())
```

```python
invalid_shipping_dates = df[df['Ship Date'] < df['Order Date']]

print("Rows where Ship Date is before Order Date:",
      len(invalid_shipping_dates))
```

```python
invalid_quantity = df[df['Quantity'] <= 0]

print("Rows with invalid Quantity:", len(invalid_quantity))
```

```python
invalid_discount = df[
    (df['Discount'] < 0) | (df['Discount'] > 1)
]

print("Rows with invalid Discount:", len(invalid_discount))
```

```python
invalid_sales = df[df['Sales'] <= 0]

print("Rows with invalid Sales:", len(invalid_sales))
```

```python
df['Postal Code'] = df['Postal Code'].astype(str)

invalid_postal = df[
    df['Postal Code'].str.len() != 5
]

print("Postal codes not having 5 digits:",
      len(invalid_postal))
```

```python
issue_log = pd.DataFrame({
    'Issue Type': [
        'Missing Values',
        'Duplicate Rows',
        'Invalid Order Dates',
        'Invalid Ship Dates',
        'Ship Date Before Order Date',
        'Invalid Quantity',
        'Invalid Discount',
        'Invalid Sales',
        'Postal Code Format'
    ],
    'Issue Count': [
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df['Order Date'].isnull().sum(),
        df['Ship Date'].isnull().sum(),
        len(invalid_shipping_dates),
        len(invalid_quantity),
        len(invalid_discount),
        len(invalid_sales),
        len(invalid_postal)
    ]
})

issue_log
```

```python
cleaned_df = df.copy()

cleaned_df['Postal Code'] = (
    cleaned_df['Postal Code']
    .astype(str)
    .str.zfill(5)
)

print("Original rows:", len(df))
print("Cleaned rows:", len(cleaned_df))
```

```python
cleaned_df.to_csv('/content/Superstore_Cleaned.csv', index=False)

print("Cleaned dataset saved successfully!")
```

```python
issue_log.to_csv('/content/Superstore_Issue_Log.csv', index=False)

print("Issue log saved successfully!")
```
